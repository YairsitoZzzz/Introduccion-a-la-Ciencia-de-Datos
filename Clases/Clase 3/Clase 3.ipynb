## 4 ¿Qué hacer con los outliers?

```python
data_no_outliers = data.copy() #hacemos una copia de los datos originiales para no alterarlos

for col in data_no_outliers.select_dtypes(include=[np.number]).columns:  
    #recorremos las columnas en nuestra copia 
    # .selec_dtypes(np) selecciona las columnas (include=[np.number]) incluye solamente columnas numericas y .columns nos da sus nombres
     #Para cada columna numérica del DataFrame, guarda temporalmente su nombre en col y ejecuta el código que aparece debajo.
    
    Q1 = data_no_outliers[col].quantile(0.25) # Primer quartile (25%)
    Q3 = data_no_outliers[col].quantile(0.75) # Tercer quartile (75%)
    IQR = Q3 - Q1 # Rango interquartile

    # limites IQR
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    # contador    , cuantos outliers por columna hay.
    outliers_count = (
        (data_no_outliers[col] < lower_bound)
        | (data_no_outliers[col] > upper_bound)
        ).sum()
    
    # apping
    # np.clip(datos,minimo,maxicmo) esto obliga a quedar entre los limites 
    data_no_outliers[col] = np.clip(data_no_outliers[col], lower_bound, upper_bound)
    
    print(f"'{col}': {outliers_count} outliers capped | Bounds: [{lower_bound:.2f}, {upper_bound:.2f}]")
 #.2f solo muestra 2 decimales
print("\nOutlier capping completed.")
#\n salta una linea
```

### llenar_faltantes.ipynb

```python
import numpy as np

def llenar_faltantes(lista):
    # Convertir la lista a un array de NumPy, reemplazando '' con NaN
    datos = np.array([float(x) if x != '' else np.nan for x in lista])

    # Recorrer todos los elementos de la lista
    for i in range(len(datos)):
        if np.isnan(datos[i]):  # Si el valor es NaN (antes era '')
            # Buscar el valor anterior más cercano
            previo = 0
            for j in range(i-1, -1, -1):  #aqui estamos llendo hacia atras recuerda (inicio , final , de cuanto en cuanto)
                if not np.isnan(datos[j]):
                    previo = datos[j]
                    break

            # Buscar el valor siguiente más cercano
            siguiente = 0
            for j in range(i+1, len(datos)):
                if not np.isnan(datos[j]):
                    siguiente = datos[j]
                    break

            # Reemplazar NaN con la media de los vecinos más cercanos
            datos[i] = (previo + siguiente) / 2

    return datos.tolist()  # Convertir el array de NumPy de vuelta a lista
```

```python
df_copia = df.copy()

# guardar las primeras 10 filas ANTES
antes_rellenar = df_copia[df_copia['AverageTemperature'].isna()].head(10).copy()
print("ANTES de rellenar valores faltantes:")
print(antes_rellenar[['City', 'AverageTemperature']])

# orden
df_copia = df_copia.sort_values(['City', 'Date']) ## ordenamos temporalmente, para que tenga sentido el anterior y despues 

# aplicar la función
for ciudad in df_copia['City'].unique():  #unique obtiene valores diferentes sin repetirlos , basicamente para cada ciudad diferente
    indice_ciudad = df_copia['City'] == ciudad
    temperaturas = df_copia.loc[indice_ciudad, 'AverageTemperature'].tolist() #seleccionamos la ciudad y obtenemos solo  la temperatura, los pasa a una lista 
    temperaturas_rellenadas = llenar_faltantes(temperaturas)
    df_copia.loc[indice_ciudad, 'AverageTemperature'] = temperaturas_rellenadas

# DESPUÉS de rellenaR
despues_rellenar = df_copia.loc[antes_rellenar.index]
print("\nDESPUÉS de rellenar valores faltantes:")
print(despues_rellenar[['City', 'AverageTemperature']])
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Lista de ciudades que pide el ejercicio
ciudades = [
    'Rome',
    'New York',
    'Bangkok',
    'Tokyo',
    'Paris',
    'Lima',
    'Cairo',
    'Moscow',
    'Peking',
    'Mexico'
]

# Seleccionamos solamente las ciudades que necesitamos
datos = df_copia[df_copia['City'].isin(ciudades)].copy()

# Convertimos la columna Date a formato de fecha
datos['Date'] = pd.to_datetime(datos['Date'])

# Creamos una nueva columna con solamente el año
datos['Year'] = datos['Date'].dt.year


# ---------------------------------------------------
# CORREGIR BANGKOK
# ---------------------------------------------------

# En el punto anterior vimos que Bangkok parece estar en Fahrenheit
# por eso la convertimos a Celsius

datos.loc[datos['City'] == 'Bangkok', 'AverageTemperature'] = (
    datos.loc[datos['City'] == 'Bangkok', 'AverageTemperature'] - 32
) * 5 / 9


# ---------------------------------------------------
# TEMPERATURA MEDIA ANUAL
# ---------------------------------------------------

# Agrupamos por ciudad y por año
# y calculamos la temperatura promedio de cada año

media_anual = datos.groupby(
    ['City', 'Year']
)['AverageTemperature'].mean().reset_index()


# ---------------------------------------------------
# GRAFICA DE TEMPERATURA MEDIA ANUAL
# ---------------------------------------------------

plt.figure(figsize=(12, 7))

for ciudad in ciudades:

    ciudad_datos = media_anual[
        media_anual['City'] == ciudad
    ]

    plt.plot(
        ciudad_datos['Year'],
        ciudad_datos['AverageTemperature'],
        label=ciudad
    )

plt.xlabel('Año')
plt.ylabel('Temperatura media')
plt.title('Temperatura media anual por ciudad')

plt.legend()

plt.show()


# ---------------------------------------------------
# PERIODOS DE 5 AÑOS
# ---------------------------------------------------

# Creamos el periodo al que pertenece cada año

media_anual['Periodo_5'] = (
    media_anual['Year'] // 5
) * 5


# Calculamos la temperatura promedio de cada periodo

media_5 = media_anual.groupby(
    ['City', 'Periodo_5']
)['AverageTemperature'].mean().reset_index()


# Grafica de periodos de 5 años

plt.figure(figsize=(12, 7))

for ciudad in ciudades:

    ciudad_datos = media_5[
        media_5['City'] == ciudad
    ]

    plt.plot(
        ciudad_datos['Periodo_5'],
        ciudad_datos['AverageTemperature'],
        label=ciudad
    )

plt.xlabel('Periodo')
plt.ylabel('Temperatura media')
plt.title('Temperatura media cada 5 años')

plt.legend()

plt.show()


# ---------------------------------------------------
# PERIODOS DE 10 AÑOS
# ---------------------------------------------------

media_anual['Periodo_10'] = (
    media_anual['Year'] // 10
) * 10


media_10 = media_anual.groupby(
    ['City', 'Periodo_10']
)['AverageTemperature'].mean().reset_index()


# Grafica de periodos de 10 años

plt.figure(figsize=(12, 7))

for ciudad in ciudades:

    ciudad_datos = media_10[
        media_10['City'] == ciudad
    ]

    plt.plot(
        ciudad_datos['Periodo_10'],
        ciudad_datos['AverageTemperature'],
        label=ciudad
    )

plt.xlabel('Periodo')
plt.ylabel('Temperatura media')
plt.title('Temperatura media cada 10 años')

plt.legend()

plt.show()


# ---------------------------------------------------
# PERIODOS DE 25 AÑOS
# ---------------------------------------------------

media_anual['Periodo_25'] = (
    media_anual['Year'] // 25
) * 25


media_25 = media_anual.groupby(
    ['City', 'Periodo_25']
)['AverageTemperature'].mean().reset_index()


# Grafica de periodos de 25 años

plt.figure(figsize=(12, 7))

for ciudad in ciudades:

    ciudad_datos = media_25[
        media_25['City'] == ciudad
    ]

    plt.plot(
        ciudad_datos['Periodo_25'],
        ciudad_datos['AverageTemperature'],
        label=ciudad
    )

plt.xlabel('Periodo')
plt.ylabel('Temperatura media')
plt.title('Temperatura media cada 25 años')

plt.legend()

plt.show()


# ---------------------------------------------------
# COMPARAR TEMPERATURAS ANTIGUAS Y RECIENTES
# ---------------------------------------------------

print("Cambio de temperatura por ciudad")
print("---------------------------------")


for ciudad in ciudades:

    ciudad_datos = media_anual[
        media_anual['City'] == ciudad
    ]

    # Temperatura promedio entre 1850 y 1874
    antigua = ciudad_datos[
        (ciudad_datos['Year'] >= 1850) &
        (ciudad_datos['Year'] <= 1874)
    ]['AverageTemperature'].mean()

    # Temperatura promedio entre 1988 y 2012
    reciente = ciudad_datos[
        (ciudad_datos['Year'] >= 1988) &
        (ciudad_datos['Year'] <= 2012)
    ]['AverageTemperature'].mean()

    # Calculamos el cambio
    cambio = reciente - antigua

    print(ciudad)
    print("Temperatura antigua:", antigua)
    print("Temperatura reciente:", reciente)
    print("Cambio:", cambio)
    print()
```
