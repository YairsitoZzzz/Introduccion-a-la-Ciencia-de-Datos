En esta clase se estudiara el conjunto de datos llamado "Wisconsin Diagnostic Breast Cancer" usando panda y se aplicaran los coeficientes de correlacion de Pearson y Spearman para encontrar correlaciones entre las diferentes variables del conjunto de datos. Ademas se usará por primera vez ' scikit-learn ' y se revisarán las primeras bases para la clasificación.

```python
import pandas as pd
from sklearn.datasets import load_breast_cancer

# Cargar el conjunto de datos
cancer = load_breast_cancer()

# Crear el DataFrame con las características
df = pd.DataFrame(
    data=cancer.data,
    columns=cancer.feature_names
)

# Agregar el diagnóstico
df['target'] = cancer.target

# Mostrar las primeras filas
print(df.head())

# Mostrar dimensiones
print(df.shape)
```
