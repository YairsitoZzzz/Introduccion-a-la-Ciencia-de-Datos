```python
import pandas as pd
x = np.linspace(-5, 5, 200) #linspace genera 200 numeros igualmente espaciados desde -5 a 5 ,
datasets = {   ## esto es un diccionario guarda-------- clave:valor 
    "Lineal": 2*x + 1,
    "Cuadratica": x**2,
    "Seno": np.sin(x),
}
for name, y in datasets.items():  # .items() permite recorrer clave y valor del diccionario|||| name ,y name guarda la clave, y guarda el valor.
    df = pd.DataFrame({"x": x, "y": y}) #se crea un data frame "x":x ----> columna x utiliza los valores de x
    pearson = df.corr().loc["x", "y"]
    spearman = df.corr(method="spearman").loc["x", "y"]
    plt.figure()
    plt.scatter(df["x"], df["y"], s=10)
    plt.title(f"P={pearson:.2f}, S={spearman:.2f}")
    plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


# -------------------------
# Datos del cuarteto
# -------------------------

x = np.array([10, 8, 13, 9, 11, 14, 6, 4, 12, 7, 5])

y1 = np.array([
    8.04, 6.95, 7.58, 8.81, 8.33,
    9.96, 7.24, 4.26, 10.84, 4.82, 5.68
])

y2 = np.array([
    9.14, 8.14, 8.74, 8.77, 9.26,
    8.10, 6.13, 3.10, 9.13, 7.26, 4.74
])

y3 = np.array([
    7.46, 6.77, 12.74, 7.11, 7.81,
    8.84, 6.08, 5.39, 8.15, 6.42, 5.73
])

x4 = np.array([
    8, 8, 8, 8, 8, 8, 8, 19, 8, 8, 8
])

y4 = np.array([
    6.58, 5.76, 7.71, 8.84, 8.47,
    7.04, 5.25, 12.50, 5.56, 7.91, 6.89
])


# -------------------------
# Guardamos los 4 conjuntos
# -------------------------

datasets = {
    "Conjunto 1": (x, y1),
    "Conjunto 2": (x, y2),
    "Conjunto 3": (x, y3),
    "Conjunto 4": (x4, y4)
}


# -------------------------
# Recorremos cada conjunto
# -------------------------

for nombre, (datos_x, datos_y) in datasets.items():

    # Creamos el DataFrame
    df = pd.DataFrame({
        "x": datos_x,
        "y": datos_y
    })

    # Calculamos las medias
    media_x = np.mean(datos_x)
    media_y = np.mean(datos_y)

    # Calculamos las varianzas
    varianza_x = np.var(datos_x)
    varianza_y = np.var(datos_y)

    # Calculamos la correlación de Pearson
    pearson = df.corr().loc["x", "y"]

    # Imprimimos los resultados
    print(nombre)
    print(f"Media de x: {media_x:.2f}")
    print(f"Media de y: {media_y:.2f}")
    print(f"Varianza de x: {varianza_x:.2f}")
    print(f"Varianza de y: {varianza_y:.2f}")
    print(f"Correlación de Pearson: {pearson:.2f}")
    print()

    # Creamos la gráfica
    plt.figure()

    # Graficamos los puntos
    plt.scatter(datos_x, datos_y)

    # Agregamos título y etiquetas
    plt.title(nombre)
    plt.xlabel("X")
    plt.ylabel("Y")

    # Mostramos la gráfica
    plt.show()
```
