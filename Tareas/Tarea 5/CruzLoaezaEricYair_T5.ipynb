#### 1 Volver a entrenar el conjunto como se hizo en la seccion 5 pero ahora con 

```python
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split

dataset = load_wine()

X = dataset["data"]
y = dataset["target"]

#los datos sin mezclar
X_train1, X_test1, y_train1, y_test1 = train_test_split(X, y, test_size=0.2, shuffle=False)

#los datos mezclados
X_train2, X_test2, y_train2, y_test2 = train_test_split(X, y, test_size=0.2, shuffle=True, random_state=42)

#los datos con estratificacion
X_train3, X_test3, y_train3, y_test3 = train_test_split(X, y, test_size=0.2, shuffle=True, stratify=y, random_state=42)

print("Sin mezclar")
print("Entrenamiento:", len(X_train1), " Prueba:", len(X_test1))

print("Mezclando")
print("Entrenamiento:", len(X_train2), "Prueba:", len(X_test2))

print("Con estratificacion")
print("Entrenamiento:", len(X_train3), "Prueba:", len(X_test3))
```

### 2. División de los conjuntos de entrenamiento y prueba
Crea los subconjuntos `X_train`, `y_train`, `X_test` y `y_test`, respectivamente. Cuenta la **cantidad** de muestras de cada clase en los conjuntos de entrenamiento y prueba. Calcula los porcentajes de cada clase en ambos conjuntos para las tres opciones de división.


```python

import pandas as pd

#sin mezclar
train1 = pd.Series(y_train1).value_counts().reindex([0, 1, 2], fill_value=0)
test1 = pd.Series(y_test1).value_counts().reindex([0, 1, 2], fill_value=0)

tabla1 = pd.DataFrame({
    "entrenamiento": train1,
    "% entrenamiento": (train1 / len(y_train1) * 100).round(2),
    "prueba": test1,
    "% prueba": (test1 / len(y_test1) * 100).round(2)
})

print("Sin mezclar")
print(tabla1)


# mezclando
train2 = pd.Series(y_train2).value_counts().reindex([0, 1, 2], fill_value=0)
test2 = pd.Series(y_test2).value_counts().reindex([0, 1, 2], fill_value=0)

tabla2 = pd.DataFrame({
    "entrenamiento": train2,
    "% entrenamiento": (train2 / len(y_train2) * 100).round(2),
    "prueba": test2,
    "% prueba": (test2 / len(y_test2) * 100).round(2)
})

print("\nMezclando")
print(tabla2)


#con estratificacion
train3 = pd.Series(y_train3).value_counts().reindex([0, 1, 2], fill_value=0)
test3 = pd.Series(y_test3).value_counts().reindex([0, 1, 2], fill_value=0)

tabla3 = pd.DataFrame({
    "entrenamiento": train3,
    "% entrenamiento": (train3 / len(y_train3) * 100).round(2),
    "prueba": test3,
    "% prueba": (test3 / len(y_test3) * 100).round(2)
})

print("\nCon estratificacion")
print(tabla3)
```

### 3. Entrenamiento y evaluación del modelo
Entrena un modelo para cada una de las tres opciones y calcula la precisión. Compara y analiza estos valores con la precisión obtenida previamente.


```python

from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

# sin mezclar
clf1 = DecisionTreeClassifier(random_state=42)
clf1.fit(X_train1, y_train1)

pred1 = clf1.predict(X_test1)
acc1 = accuracy_score(y_test1, pred1)

print("sin mezclar:", round(acc1 * 100, 2), "%")


#mezclando
clf2 = DecisionTreeClassifier(random_state=42)
clf2.fit(X_train2, y_train2)

pred2 = clf2.predict(X_test2)
acc2 = accuracy_score(y_test2, pred2)

print("mezclando:", round(acc2 * 100, 2), "%")


#con estratificacion
clf3 = DecisionTreeClassifier(random_state=42)
clf3.fit(X_train3, y_train3)

pred3 = clf3.predict(X_test3)
acc3 = accuracy_score(y_test3, pred3)

print("con estratificacion:", round(acc3 * 100, 2), "%")
```

#### 4. Evaluación de métricas
Estudia también las métricas: *precision*, *recall* y *F1-score*.


```python
from sklearn.metrics import classification_report

print("sin mezclar")
print(classification_report(y_test1, pred1, labels=[0, 1, 2], zero_division=0))

print("mezclando")
print(classification_report(y_test2, pred2, labels=[0, 1, 2], zero_division=0))

print("con estratificacion")
print(classification_report(y_test3, pred3, labels=[0, 1, 2], zero_division=0))
```

#### 5. Análisis de resultados
¿Cuál es el valor más razonable? ¿Por qué? ¿En qué casos es crítico utilizar `stratify=y` y cuándo no es tan importante?


El valor mas razonable es el 94.4%  qe obtuvimos con la estratificacion por que los lados de entranimeinto y prueba mantienen poncentajes parecidos de las tres clases.  
Usar stratify es importante cuando tenemos muchas muestras de una clase y pocas de otra ya que al divividie datos podriamos dejar alguna chlase con muy pocas muestras de entrenamiento, por el contratrio si tenemos sufientes datos  no estan necesario ya que al mezclar los datos normalmente se obtienen proporciones similares.
